```python
import pandas as pd
```

```python
customers=pd.read_csv(r"C:\Users\qscaxz\Downloads\customers_clean.csv")
loans=pd.read_csv(r"C:\Users\qscaxz\Downloads\loans_clean.csv")
collections=pd.read_csv(r"C:\Users\qscaxz\Downloads\collections_status_clean.csv")
```

```python
df=loans.merge(customers, on='Customer_ID').merge(collections, on='Loan_ID')
```

```python
df.head()
```

```python
df['Priority_Score']=((df['Outstanding_Amount']/1000) +(df['Days_Past_Due']*2)+ (df['Missed_EMIs']*10))
```

```python
df['Priority_Score']+=df['Contact_Response'].map(
    {
        'No Answer': 30,
        'Promise to Pay': 15,
        'Responsive': 5
    }
)
```

```python
def risk_bucket(score):
    if score >=250:
        return "High Risk"
    elif score<=120:
        return "Medium Risk"
    else:
        return "Low Risk"
```

```python
df['Risk_Level']=df['Priority_Score'].apply(risk_bucket)
```

```python
df.to_csv(r"C:\Users\qscaxz\Downloads\analytics_dataset.csv", index=False)
```
