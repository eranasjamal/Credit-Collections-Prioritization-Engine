```python
customers_clean=customers_raw.copy()
```

```python
customers_clean.head()
```

```python
customers_clean['City'].value_counts()
```

```python
customers_clean['City']=customers_clean['City'].str.strip().str.title()
```

```python
customers_clean['City']=customers_clean['City'].replace({"Delhi Ncr":"Delhi NCR"})
```

```python
customers_clean["Income_Band"]=customers_clean["Income_Band"].fillna('Medium')
```

```python
customers_clean.isnull().sum()
```

```python
customers_clean['Credit_Score'] = customers_clean['Credit_Score'].fillna(
    customers_clean['Credit_Score'].median()
)
```

```python
customers_clean=customers_clean.drop_duplicates()
```

```python
customers_clean.duplicated().sum()
```

```python
customers_clean['Credit_Score']=customers_clean['Credit_Score'].fillna(customers_clean['Credit_Score'].median())
```

```python
customers_clean.shape
```

```python
customers_clean.to_csv('customers_clean.csv', index=False)
```
